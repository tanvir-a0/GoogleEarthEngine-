```python
bangladesh_districts = ee.FeatureCollection("FAO/GAUL/2015/level2") \
    .filter(ee.Filter.eq('ADM0_NAME', 'Bangladesh'))
```

```python
def calculate_growth(feature):
    name = feature.get('ADM2_NAME')
    p2011_val = ee.Number(pop2011.get(name))
    p2022_val = ee.Number(pop2022.get(name))
    
    #growth formula: ((new - old) / old) * 100
    growth = p2022_val.subtract(p2011_val).divide(p2011_val).multiply(100)
    
    #returning data
    return feature.set('growth_rate', growth).set('pop_2022', p2022_val)

growth_map = bangladesh_districts.map(calculate_growth)

#base image
growth_image = growth_map.reduceToImage(
    properties=['growth_rate'],
    reducer=ee.Reducer.first()
)

pos_growth = growth_image.updateMask(growth_image.gt(0))
pos_vis = {
    'min': 0, 
    'max': 50, 
    'palette': ['ffffff', 'ffcccc', 'ff6666', 'ff0000', '800000']
}

neg_growth = growth_image.updateMask(growth_image.lt(0))
neg_vis = {
    'min': -10, 
    'max': 0,
    'palette': ['006400', '32CD32', 'ffffff']
}

```

```python
m = geemap.Map()
m.centerObject(bangladesh_districts, 7)
m.set_options('HYBRID')


m.addLayer(pos_growth, pos_vis, 'Positive Growth')
m.addLayer(neg_growth, neg_vis, 'Negative Growth')

#adding borders
borders = ee.Image().paint(growth_map, 0, 1)
m.addLayer(borders, {'palette': '000000'}, 'bordars')

legend_dict = {
    "Positive Growth": "ff0000",
    "Negative Growth": "008000" 
}
m.add_legend(title="Growth Rate (between 2011 and 2022)", legend_dict=legend_dict, position='bottomright')

m
```
