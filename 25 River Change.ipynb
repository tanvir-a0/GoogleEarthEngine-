```python
import ee
import geemap
from dotenv import load_dotenv
import os
import ipywidgets as widgets
from datetime import datetime
import matplotlib.pyplot as plt
import pandas as pd

load_dotenv(dotenv_path='.env')
project_name = os.getenv("project_name")
# print(f"Project Name: {project_name}")
ee.Initialize(project=project_name)
```

```python
Map = geemap.Map()

country = ee.FeatureCollection("USDOS/LSIB_SIMPLE/2017") \
    .filter(ee.Filter.eq('country_na', 'Bangladesh'))

Map.centerObject(country, 7)
Map.add_basemap('SATELLITE')
Map
```

```python
jrc = ee.Image("JRC/GSW1_4/GlobalSurfaceWater")
#jrc gives us water occurrence data over time

change = jrc.select('transition')
palette_list = [
    'ffffff', # 0 No change
    '0000ff', # 1 Permanent
    '22b14c', # 2 New permanent (Land -> Water)
    'd1102d', # 3 Lost permanent (Water -> Land)
    '99d9ea', # 4 Seasonal
    'b5e61d', # 5 New seasonal
    'e6a1aa', # 6 Lost seasonal
    'ff7f27', # 7 Seasonal -> permanent
    'ffc90e', # 8 Permanent -> seasonal
    '7f7f7f', # 9 Ephemeral permanent
    'c3c3c3'  # 10 Ephemeral seasonal
]

transition_vis = {
    'bands': ['transition'],
    'min': 0,
    'max': 10,
    'palette': palette_list
}

Map.addLayer(change, transition_vis, 'River Shifting History')
print("data info: ",  jrc.propertyNames().getInfo())
print("time start:", ee.Date(jrc.get('system:time_start')).format('YYYY-MM-dd').getInfo())
print("time end:", ee.Date(jrc.get('system:time_end')).format('YYYY-MM-dd').getInfo())
Map
```

```python
#adding colored legend
legend_dict = {
    "No Change": "#ffffff",
    "Permanent Water": "#0000ff",
    "New Permanent Water (Land -> Water)": "#22b14c",
    "Lost Permanent Water (Water -> Land)": "#d1102d",
    "Seasonal Water": "#99d9ea",
    "New Seasonal Water": "#b5e61d",
    "Lost Seasonal Water": "#e6a1aa",
    "Seasonal to Permanent": "#ff7f27",
    "Permanent to Seasonal": "#ffc90e",
    "Ephemeral Permanent Water": "#7f7f7f",
    "Ephemeral Seasonal Water": "#c3c3c3"
}

Map.add_legend(title="River Change Legend", legend_dict=legend_dict, position='bottomright')

Map
```
