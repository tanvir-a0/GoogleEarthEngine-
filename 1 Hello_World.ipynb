```python
import ee
import geemap
from dotenv import load_dotenv
import os
load_dotenv(dotenv_path='.env')
project_name = os.getenv("project_name")
# print(f"Project Name: {project_name}")
ee.Initialize(project=project_name)
```

```python
map = geemap.Map()

dem = ee.Image('USGS/SRTMGL1_003') #digital elevation model

vis_params = {
    "min": 0,
    "max": 1000, # max elevation in meters
    'palette': ['006633', 'E5FFCC', '662A00', 'D8D8D8', 'F5F5F5']
}

```

```python
map.setCenter(91.8349, 22.3752, 10) #longitude, latitude, zoom level
map.addLayer(dem, vis_params, 'SRTM') #dem layer, visualization parameters, layer name
map
```
